## 國中數學：基本運算

### 1. 四則運算

Python 的加、減、乘、除與國中學的四則運算完全一樣，而且同樣遵守「先乘除、後加減」的**運算子優先順序**。



````python
print(3 + 5)          # 加法 → 8
print(10 - 4)         # 減法 → 6
print(3 * 5)          # 乘法 → 15
print(15 / 4)         # 除法 → 3.75（永遠回傳浮點數）
print(15 // 4)        # 整數除法 → 3（只取商的整數部分）
print(15 % 4)         # 餘數運算 → 3（15 ÷ 4 的餘數）
print(2 ** 10)        # 次方運算 → 1024（2 的 10 次方）
````


**詳細說明：**

- `+`、`-`、`*`、`/` 分別對應加、減、乘、除。
- `//` 是**整數除法**，只留下商數，不考慮餘數。
- `%` 是**取餘數**，例如 `15 % 4 = 3`，因為 `15 ÷ 4 = 3 ... 3`。
- `**` 是**次方**，例如 `2 ** 10` 就是「2 的 10 次方」。
- 可以用括號 `()` 改變運算順序：



````python
print(3 + 5 * 2)      # 先乘後加 → 13
print((3 + 5) * 2)    # 先加再乘 → 16
````


### 2. 判斷奇數與偶數

國中常考的題目：判斷一個數是奇數還是偶數，可以用「除以 2 的餘數」來判斷。



````python
number = 17
if number % 2 == 0:
    print(number, "是偶數")
else:
    print(number, "是奇數")
# 輸出：17 是奇數
````


**詳細說明：**

- 偶數除以 2 的餘數是 0，奇數除以 2 的餘數是 1。
- `if ... else ...` 是條件判斷，當條件成立時執行 `if` 底下的程式，否則執行 `else` 底下的程式。

### 3. 解一元一次方程式

例如：`3x + 5 = 20`，移項可得 `x = (20 - 5) / 3`。用 Python 可以直接算出答案。



````python
# 解 3x + 5 = 20
# 移項：3x = 20 - 5，x = (20 - 5) / 3
x = (20 - 5) / 3
print("x =", x)        # 輸出：x = 5.0
````


### 4. 兩點距離公式

在座標平面上，兩點 `(x1, y1)` 與 `(x2, y2)` 的距離公式為：



```距離 = √((x2 - x1)² + (y2 - y1)²)
```

````python
import math

x1, y1 = 1, 2
x2, y2 = 4, 6
distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
print("兩點距離 =", distance)   # 輸出：兩點距離 = 5.0
````


**詳細說明：**

- `math.sqrt()` 是開根號（平方根）函式。
- `(x2 - x1) ** 2` 就是 `(x2 - x1)` 的平方。
- 這就是國中學的「畢氏定理」應用：直角三角形的斜邊長。

## 國中數學：分數與根號

### 5. 使用分數運算（fractions 模組）

電腦計算小數時可能出現誤差，例如 `0.1 + 0.2` 並不等於 `0.3`。要精確計算分數，可以使用 `fractions.Fraction`。



````python
from fractions import Fraction

a = Fraction(1, 3)      # 三分之一
b = Fraction(1, 6)      # 六分之一
print(a + b)            # 1/3 + 1/6 = 1/2
print(a * b)            # 1/3 × 1/6 = 1/18
````


**詳細說明：**

- `Fraction(分子, 分母)` 可以建立精確的分數，不會有小數誤差。
- 分數的加減乘除都可以直接用運算子完成。
- `1/3 + 1/6 = 2/6 + 1/6 = 3/6 = 1/2`，程式的輸出就是精確的 `1/2`。

### 6. 最大公因數與最小公倍數



````python
import math

print(math.gcd(12, 18))     # 最大公因數 → 6
print(math.lcm(4, 6))       # 最小公倍數 → 12
````


**詳細說明：**

- `math.gcd(a, b)` 回傳 `a` 和 `b` 的**最大公因數**（Greatest Common Divisor）。
- `math.lcm(a, b)` 回傳 `a` 和 `b` 的**最小公倍數**（Least Common Multiple）。
- 約分、通分、分數加減都需要用到這兩個概念。

### 7. 質數判斷與質因數分解



````python
# 判斷一個數是否為質數
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):   # 只需檢查到根號 n
        if n % i == 0:
            return False
    return True

print(is_prime(17))     # 輸出：True（17 是質數）
print(is_prime(15))     # 輸出：False（15 = 3 × 5）
````


**詳細說明：**

- 質數是大於 1 且只能被 1 和自己整除的數。
- 檢查時只需要試除到 `√n` 就好，因為若 `n` 有因數，必然有一個因數小於等於 `√n`。
- `range(2, int(n ** 0.5) + 1)` 產生從 2 到 `√n` 的整數序列，`int()` 把浮點數轉成整數。

## 高中數學：代數與函數

### 8. 多項式運算



````python
# 多項式 f(x) = 3x² + 2x + 1，求 f(2)
def f(x):
    return 3 * x ** 2 + 2 * x + 1

print(f(2))     # 3×4 + 2×2 + 1 = 17
````


**詳細說明：**

- 定義函式 `f(x)`，Python 會依照公式自動代入計算。
- `x ** 2` 代表 `x²`。

### 9. 求等差數列與等比數列

**等差數列：** 第一項 `a1`，公差 `d`，第 n 項公式 `an = a1 + (n - 1) × d`，前 n 項和 `Sn = n × (a1 + an) / 2`。



````python
# 等差數列：2, 5, 8, 11, ...（首項 2，公差 3）
a1, d = 2, 3
n = 10
an = a1 + (n - 1) * d
Sn = n * (a1 + an) / 2
print("第10項 =", an)      # 29
print("前10項總和 =", Sn)  # 155
````


**等比數列：** 首項 `a1`，公比 `r`，第 n 項公式 `an = a1 × r^(n-1)`。



````python
# 等比數列：2, 6, 18, 54, ...（首項 2，公比 3）
a1, r = 2, 3
n = 5
an = a1 * r ** (n - 1)
print("第5項 =", an)      # 162
````


**詳細說明：**

- 等差數列：相鄰兩項的差固定，用加法表示。
- 等比數列：相鄰兩項的比固定，用乘法（次方）表示。

### 10. 求等差級數總和（高斯小故事）

數學家高斯小時候的題目：`1 + 2 + 3 + ... + 100 = ?`



````python
n = 100
total = n * (n + 1) // 2
print("1 加到 100 =", total)      # 5050
````


**詳細說明：**

- 這個公式由高斯發現：把第一項和最後一項相加 `(1 + 100)`，第二項和倒數第二項相加 `(2 + 99)`，每一組都等於 101，共有 50 組，所以 `50 × 101 = 5050`。
- 公式為 `n × (n + 1) / 2`，也就是等差級數總和公式的特例。

### 11. 二次函數的頂點與最大最小值

二次函數 `y = ax² + bx + c`，頂點 x 座標為 `x = -b / (2a)`。



````python
# y = x² - 4x + 3，a=1, b=-4, c=3
a, b, c = 1, -4, 3
vertex_x = -b / (2 * a)
vertex_y = a * vertex_x ** 2 + b * vertex_x + c
print("頂點 =", (vertex_x, vertex_y))   # 頂點 = (2.0, -1.0)
````


**詳細說明：**

- 二次函數的圖形是拋物線，頂點就是最高點或最低點。
- 因為 `a = 1 > 0`，開口向上，頂點 `(2, -1)` 就是最小值所在。
- `-b / (2a)` 的推導來自配方法，高中課程會詳細講解。

## 高中數學：三角函數、指數與對數

### 12. 三角函數

Python 的 `math` 模組提供 `sin`、`cos`、`tan` 等函式，**注意角度要用弧度（radian）**，不是度（degree）。



````python
import math

# 角度轉弧度
angle_deg = 30
angle_rad = math.radians(angle_deg)      # 度 → 弧度

print(math.sin(angle_rad))      # sin 30° = 0.5
print(math.cos(angle_rad))      # cos 30° ≈ 0.866
print(math.tan(angle_rad))      # tan 30° ≈ 0.577
print(math.degrees(math.pi))    # π 弧度 = 180 度
````


**詳細說明：**

- `math.radians(度)` 把角度轉成弧度。
- `math.degrees(弧度)` 把弧度轉成角度。
- 常用特殊角：`sin 30° = 1/2`、`cos 60° = 1/2`、`tan 45° = 1`。
- 進階函式還有：`math.asin()`（反正弦）、`math.acos()`、`math.atan()`。

### 13. 指數與對數



````python
import math

# 指數運算
print(2 ** 3)               # 2³ = 8
print(10 ** 5)              # 10⁵ = 100000

# 對數運算
print(math.log(8, 2))       # log₂8 = 3（以 2 為底）
print(math.log(1000, 10))   # log₁₀1000 = 3
print(math.log10(1000))     # 同上的簡寫 → 3.0
print(math.log(math.e))     # ln e = 1（自然對數，底數 e）
````


**詳細說明：**

- `math.log(x, 底數)` 計算以指定底數為底的對數。
- `math.log10(x)` 是以 10 為底的常用對數。
- `math.log(x)` 預設是以 `e` 為底的自然對數 `ln`。
- 對數與指數互為反函數：若 `2³ = 8`，則 `log₂8 = 3`。

### 14. 一元二次方程式的解（公式解）

方程式 `ax² + bx + c = 0` 的解為：



```x = (-b ± √(b² - 4ac)) / (2a)
```

````python
import math

def solve_quadratic(a, b, c):
    discriminant = b ** 2 - 4 * a * c     # 判別式
    if discriminant > 0:
        x1 = (-b + math.sqrt(discriminant)) / (2 * a)
        x2 = (-b - math.sqrt(discriminant)) / (2 * a)
        return (x1, x2)                   # 兩相異實根
    elif discriminant == 0:
        x = -b / (2 * a)
        return (x, x)                     # 重根
    else:
        return "無實數解"                  # 兩虛根

print(solve_quadratic(1, -5, 6))   # x² - 5x + 6 = 0 → (3.0, 2.0)
print(solve_quadratic(1, 2, 1))    # x² + 2x + 1 = 0 → (-1.0, -1.0)
print(solve_quadratic(1, 0, 4))    # x² + 4 = 0 → 無實數解
````


**詳細說明：**

- `b² - 4ac` 叫做**判別式**，決定解的形式：
  - 大於 0：有兩個相異實根。
  - 等於 0：有一個重根（兩根相同）。
  - 小於 0：沒有實數解（有兩個虛數解）。
- 這正是高中數學「二次方程式公式解」的應用。

## 大學數學：微積分

### 15. 極限與微分（數值方法）

先用「差商」的數值方法近似導數，這是最直接理解「導數是瞬間變化率」的方式：



````python
def derivative(f, x, h=1e-6):
    # 導數定義：f'(x) ≈ (f(x + h) - f(x)) / h
    return (f(x + h) - f(x)) / h

# f(x) = x²，f'(x) = 2x，所以 f'(3) 應該等於 6
f = lambda x: x ** 2
print(derivative(f, 3))      # 約 6.000001，非常接近 6
````


**詳細說明：**

- 導數的定義是「平均變化率在 `h → 0` 時的極限」：`f'(x) = lim (f(x+h) - f(x)) / h`。
- 只要把 `h` 設得非常小（例如 `1e-6`），就可以得到很接近真值的近似導數。
- `lambda x: x ** 2` 是「匿名函式」，一行就能定義簡單的函式。

### 16. 積分（數值方法：辛普森法則）

積分代表函數曲線下的面積。利用「辛普森法則」可以數值計算定積分：



````python
def simpson(f, a, b, n=1000):
    # 辛普森法則近似定積分 ∫ₐᵇ f(x) dx
    if n % 2 == 1:
        n += 1                      # n 必須是偶數
    h = (b - a) / n
    total = f(a) + f(b)
    for i in range(1, n):
        coef = 4 if i % 2 == 1 else 2
        total += coef * f(a + i * h)
    return total * h / 3

# ∫₀¹ x² dx = 1/3 ≈ 0.3333
f = lambda x: x ** 2
print(simpson(f, 0, 1))      # 約 0.3333333333333333
````


**詳細說明：**

- 定積分 `∫₀¹ x² dx = [x³/3]₀¹ = 1/3`，正確答案是 `1/3 ≈ 0.3333`。
- 辛普森法則把區間切成很多小段，用二次函數逼近每一段，是常用的數值積分方法。
- 段數 `n` 愈多，結果愈精確。

### 17. 泰勒展開（用函式逼近）

大學微積分的核心概念：任何光滑函式都可以用多項式無限逼近。例如：



```eˣ = 1 + x + x²/2! + x³/3! + x⁴/4! + ...
```

````python
import math

def taylor_exp(x, terms=10):
    # 用泰勒展開逼近 eˣ
    total = 0
    for n in range(terms):
        total += x ** n / math.factorial(n)
    return total

print(taylor_exp(1))          # 逼近 e¹，約 2.718281801
print(math.e)                 # 真正的 e = 2.718281828
````


**詳細說明：**

- `math.factorial(n)` 計算 `n!`（n 的階乘）。
- 展開項數愈多，逼近愈精確。這是電腦計算指數、三角函數等數值的基礎原理。
- 例如 `sin x = x - x³/3! + x⁵/5! - ...`，也是同樣的道理。

### 18. 用 sympy 做符號運算（進階）

如果希望 Python 直接算出「符號解」而不只是數值，可以使用 `sympy` 套件。**第一次使用需先安裝**：



```bashuv pip install sympy
```

````python
import sympy as sp

x = sp.Symbol("x")

# 微分：對 x 微分
print(sp.diff(x ** 3 + 2 * x, x))          # 3*x**2 + 2

# 積分：不定積分
print(sp.integrate(x ** 2, x))             # x**3/3

# 定積分 ∫₀¹ x² dx
print(sp.integrate(x ** 2, (x, 0, 1)))     # 1/3

# 解方程式 x² - 5x + 6 = 0
print(sp.solve(x ** 2 - 5 * x + 6, x))     # [2, 3]
````


**詳細說明：**

- `sympy`（symbolic python）可以進行**符號運算**，就像用手寫推導數學一樣。
- `sp.diff(函式, 變數)` 求導數，`sp.integrate()` 求積分，`sp.solve()` 解方程式。
- 它會回傳精確的符號答案（例如 `1/3` 而非 `0.3333`），是學大學數學的利器。

## 大學數學：線性代數

### 19. 矩陣運算

Python 內建的 `list` 可以用巢狀結構表示矩陣。這裡示範矩陣的加法與乘法：



````python
def mat_add(A, B):
    # 矩陣加法：對應位置相加
    rows = len(A)
    cols = len(A[0])
    return [[A[i][j] + B[i][j] for j in range(cols)] for i in range(rows)]

def mat_mul(A, B):
    # 矩陣乘法：A 的行數必須等於 B 的列數
    rows = len(A)
    cols = len(B[0])
    k = len(B)
    return [[sum(A[i][t] * B[t][j] for t in range(k))
             for j in range(cols)] for i in range(rows)]

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

print(mat_add(A, B))     # [[6, 8], [10, 12]]
print(mat_mul(A, B))     # [[19, 22], [43, 50]]
````


**詳細說明：**

- 矩陣加法：兩個相同大小的矩陣，對應位置相加。
- 矩陣乘法：`C[i][j] = Σ A[i][t] × B[t][j]`，也就是「左矩陣的列」與「右矩陣的行」做內積。
- 若用 `numpy` 套件，可以更簡單：



````python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(A + B)        # 矩陣加法
print(A @ B)        # 矩陣乘法（@ 是矩陣乘法運算子）
print(A.T)          # 轉置矩陣
print(np.linalg.det(A))       # 行列式值
print(np.linalg.inv(A))       # 反矩陣
````


### 20. 解線性方程組

例如解方程組：



``` x + 2y = 5
3x + 4y = 11
```


用矩陣表示為 `AX = B`，其中 `A = [[1,2],[3,4]]`、`B = [5, 11]`，解就是 `X = A⁻¹B`。



````python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([5, 11])

solution = np.linalg.solve(A, B)
print(solution)      # [1. 2.]，代表 x = 1, y = 2
````


**詳細說明：**

- `np.linalg.solve(A, B)` 直接解出線性方程組，回傳的是 `x` 和 `y` 的數值。
- 驗算：`1 + 2×2 = 5`，`3×1 + 4×2 = 11`，答案正確。
- 這是高中「克拉瑪法則」與大學「線性代數」在電腦中的實際應用。

## 大學數學：統計與機率

### 21. 平均數、中位數與標準差



````python
import statistics

data = [60, 70, 70, 80, 90, 100]

print(statistics.mean(data))      # 平均數 → 78.33...
print(statistics.median(data))    # 中位數 → 75.0
print(statistics.mode(data))      # 眾數 → 70
print(statistics.stdev(data))     # 樣本標準差（分母 n-1）
print(statistics.pstdev(data))    # 母體標準差（分母 n）
````


**詳細說明：**

- **平均數**：全部加總除以個數。
- **中位數**：排序後正中間的值；偶數個時取中間兩數的平均。
- **眾數**：出現次數最多的值。
- **標準差**：衡量資料的分散程度。`stdev` 用樣本標準差（分母 `n-1`），`pstdev` 用母體標準差（分母 `n`）。

### 22. 排列與組合



````python
import math

# 排列 P(n, r) = n! / (n-r)!
# 組合 C(n, r) = n! / (r! × (n-r)!)

def perm(n, r):
    return math.factorial(n) // math.factorial(n - r)

def comb(n, r):
    return math.factorial(n) // (math.factorial(r) * math.factorial(n - r))

print(perm(5, 3))     # 從 5 個東西選 3 個排列 → 60
print(comb(5, 3))     # 從 5 個東西選 3 個組合 → 10
````


**詳細說明：**

- **排列（Permutation）**：選取之後有順序之分，`P(5,3) = 5×4×3 = 60`。
- **組合（Combination）**：選取之後沒有順序之分，`C(5,3) = 10`。
- 關係式：`C(n, r) = P(n, r) / r!`，因為組合數等於排列數除以內部排列數。
- Python 3.8 以上也可直接使用 `math.comb(5, 3)` 與 `math.perm(5, 3)`。

### 23. 常態分配與機率



````python
import math

def normal_pdf(x, mu=0, sigma=1):
    # 常態分配機率密度函數 f(x)
    return (1 / (sigma * math.sqrt(2 * math.pi))) * math.exp(
        -((x - mu) ** 2) / (2 * sigma ** 2)
    )

# 標準常態分配 N(0,1)，x=0 時機率密度最大
print(normal_pdf(0))      # 約 0.3989
print(normal_pdf(1))      # 約 0.2420
````


**詳細說明：**

- 常態分配是統計學最重要的分配，圖形是左右對稱的鐘形曲線。
- 機率密度函式 `f(x) = (1/(σ√(2π))) · e^((x-μ)²/(2σ²))`，其中的參數 `μ`（平均數）決定曲線中心位置，`σ`（標準差）決定曲線寬窄。
- 在 `x = μ` 時，機率密度最高。平均值附近出現的機率最大。

### 24. 隨機抽樣與模擬（蒙地卡羅法估計圓周率）

蒙地卡羅法用「隨機丟點」來估算數值。在邊長為 2 的正方形內隨機丟點，計算落在半徑 1 的圓內的比例，就可以逼近圓周率：



````python
import random

def estimate_pi(n=100000):
    inside = 0
    for _ in range(n):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x * x + y * y <= 1:
            inside += 1
    return 4 * inside / n

print(estimate_pi())      # 大約 3.14 附近（會隨機浮動）
````


**詳細說明：**

- 圓面積與正方形面積的比例是 `π/4`（圓半徑 1，面積 `π`；正方形邊長 2，面積 4）。
- 點落在圓內的機率約等於 `π/4`，所以 `π ≈ 4 × (落在圓內的比例)`。
- 丟的點數愈多，估計愈精確。這是「蒙地卡羅模擬」的經典範例。
